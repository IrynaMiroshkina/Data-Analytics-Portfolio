# **Завдання 9. Аналіз популярності Python за віковими категоріями**

Опис алгоритму:


1.   Маркування цільової аудиторії: Створимо в датасеті новий стовпець - мітку, що позначає чи є респондент «Python-розробником». Використаємо раніше створену серію worked_with_python.
2.   Дослідження категорій віку: Переглянемо усі унікальні варіанти відповідей зі стовпця Age. Це необхідно, щоб побачити, як саме сформовані вікові групи в опитуванні та чи є серед них такі, що заважатимуть аналізу. Створюємо список вікових груп.
3.   Очищення критеріїв: Видалимо з датасету респондентів, що мають не релевантне значення в стовпці Age, тобто їх вік поза списком вікових груп.
4.   Агрегація та представлення результатів: Групуємо дані за віковими категоріями Age та обчислюємо середнє значення для стовпця з міткою «Python-розробник». Оскільки цей стовпець містить логічні значення, середня величина автоматично стає часткою користувачів Python у кожній групі. Переведемо частки у відсотки.

```python
# Додамо стовпець, що ідентифікує респондентів - Python-розробників
df['PythonDeveloper'] = worked_with_python
```

```python
# Переглянемо, які значення є в стовпці 'Age'
age_groups = list(df['Age'].unique())
print(age_groups)
```

```python
# Видалимо нерелевантне значення віку
age_groups.remove('Prefer not to say')
print(age_groups)
```

```python
# Відфільтруємо відповіді з відомим віком
age_known = df['Age'].isin(age_groups)
respondents_of_known_age = df.loc[age_known,['ResponseId', 'Age','PythonDeveloper']]
respondents_of_known_age.head(10)
```

```python
age_of_python_developers = (respondents_of_known_age.groupby('Age')['PythonDeveloper'].mean() * 100).round(0).reset_index()
age_of_python_developers.columns = ['Age','Percentage of respondents']
age_of_python_developers.style.format({'Percentage of respondents': '{:,.0f}%'})
```

```python
from matplotlib.lines import Line2D
# 1. Підготовка даних
age_totals = respondents_of_known_age.groupby('Age').size()
python_counts = respondents_of_known_age.groupby(['Age', 'PythonDeveloper']).size().unstack(fill_value=0)
python_counts = python_counts.reindex(columns=[True, False], fill_value=0)

labels_age = age_totals.index
sizes_outer = age_totals.values
sizes_inner = python_counts.values.flatten()

# 2. Формування підписів для внутрішнього кільця
inner_labels = []
for i in range(len(python_counts)):
    row = python_counts.iloc[i]
    total_in_group = row.sum()
    if total_in_group > 0:
        py_percent = int(round(row[True] / total_in_group * 100, 0))
        inner_labels.append(f"{py_percent}%")
    else:
        inner_labels.append("")
    inner_labels.append("")

# 3. Налаштування кольорів
colors_outer = plt.cm.Paired(np.arange(len(labels_age)))
colors_inner = []
for c in colors_outer:
    colors_inner.extend([c * 0.9, c * 0.65])

# 4. Побудова графіка
fig, ax = plt.subplots(figsize=(10, 10))

# Зовнішнє кільце
ax.pie(sizes_outer, labels=labels_age, radius=1, colors=colors_outer,
       wedgeprops=dict(width=0.3, edgecolor='w'),
       pctdistance=0.85, autopct='%1.0f%%',
       textprops={'fontsize': 11, 'fontweight': 'bold'})

# Внутрішнє кільце
ax.pie(sizes_inner, labels=inner_labels, radius=0.7, colors=colors_inner,
       wedgeprops=dict(width=0.3, edgecolor='w'),
       labeldistance=0.75,
       textprops={'fontsize': 10, 'color': 'white', 'weight': 'bold', 'ha': 'center'})

# Напис у центрі
ax.text(0, 0, 'Python-\nрозробники', ha='center', va='center',
        fontsize=14, fontweight='bold', color='#2c3e50')

ax.set(aspect="equal")
plt.title('Аналіз респондентів: розподіл за віком та популярність Python', pad=5, fontsize=15)

# Вручну налаштовуємо відступи: top=0.95 піднімає графік вище
# bottom=0.2 звільняє місце для легенди внизу
plt.subplots_adjust(top=0.95, bottom=0.15)
# ---------------------------

# 5. Легенда
legend_elements = [
    Line2D([0], [0], color='gray', alpha=0.6, lw=8, label='Сектор Python-розробників (відсоток у групі)'),
    Line2D([0], [0], color='gray', alpha=0.9, lw=8, label='Інші мови програмування')
]
ax.legend(handles=legend_elements, loc='lower center', bbox_to_anchor=(0.5, -0.05), ncol=1)

plt.show()
```

# **Завдання 10. Аналіз індустрій серед високооплачуваних віддалених працівників**

```python
# Визначимо порогове значення 75-го перцентиля за компенсацією для всього набору даних
# це сума, яку заробляє більше ніж 75% людей
percentile_75 = df['ConvertedCompYearly'].dropna().quantile(0.75)
# Створюємо маски для фільтрації:
# - працюють віддалено ('Remote')
# - зарплата >= поріг 75-го перцентиля
remote = df['RemoteWork'] == 'Remote'
high_comp = df['ConvertedCompYearly'] >= percentile_75
# Відбираємо респондентів та потрібні стовпці
top_remote_comp = df.loc[remote & high_comp, ['ResponseId', 'Industry', 'ConvertedCompYearly']]

# Виведемо результат та поріг для перевірки
print(f"Поріг 75-го перцентиля: ${percentile_75:,.0f}")
top_remote_comp.head()
```

Щоб визначити найпопулярніші індустрії для групи високооплачуваних віддалених працівників (вище 75-ого перцентиля), потрібно згрупувати відфільтровані дані за стовпцем Industry та підрахувати кількість входжень.

```python
# 1. Рахуємо частоту кожної індустрії у відфільтрованій групі
industry_counts = top_remote_comp['Industry'].value_counts().reset_index()
industry_counts.columns = ['Industry', 'Count']

# 2. Додаємо відсоток, щоб розуміти частку кожної індустрії в цій групі
total_count = len(top_remote_comp)
industry_counts['Percentage'] = (industry_counts['Count'] / total_count * 100).round(1)

# 3. Виводимо топ-10 індустрій
print("Найприбутковіші індустрії для Remote-роботи:")
industry_counts
```

```python
colors = plt.cm.tab20(np.arange(len(industry_counts)))

fig, ax = plt.subplots(figsize=(12, 10))
wedges, texts, autotexts = ax.pie(
    industry_counts['Count'],
    labels=None,                 # Назви залишаємо в легенді для чистоти
    autopct='%1.0f%%',          # Відображаємо цілі відсотки
    startangle=140,
    colors=colors,
    pctdistance=0.8,             # Відсотки ближче до краю
    explode=[0.02] * len(industry_counts) # Легке розмежування секторів
)

# Стилізація тексту відсотків (білий колір на кольорових секторах)
plt.setp(autotexts, size=10, weight="bold", color="white")
# Використовуємо subplots_adjust, щоб підняти графік до назви
plt.title('Розподіл усіх індустрій серед високооплачуваних Remote-фахівців', pad=20, fontsize=16)
# Легенда праворуч
ax.legend(
    wedges,
    industry_counts['Industry'],
    title="Індустрії",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1),
    fontsize=10
)
# Піднімаємо діаграму вище до заголовка
plt.subplots_adjust(top=0.95, bottom=0.1)
plt.axis('equal')
plt.show()
```
